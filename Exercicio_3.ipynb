```python
import matplotlib.pyplot as plt
import numpy as np

mes = np.arange(12) + 1 # meses de 1 a 12

# quantidades de produtos vendidos por mês. As entradas são de Janeiro a Dezembro
creme_facial = np.array([ 2500, 2630, 2140, 3400, 3600, 2760, 2980, 3700, 3540, 1990, 2340, 2900 ])
limpeza_facial = np.array([ 1500, 1200, 1340, 1130, 1740, 1555, 1120, 1400, 1780, 1890, 2100, 1760 ])
pasta_dentaria = np.array([ 5200, 5100, 4550, 5870, 4560, 4890, 4780, 5860, 6100, 8300, 7300, 7400 ])
sabonete = np.array([ 9200, 6100, 9550, 8870, 7760, 7490, 8980, 9960, 8100, 10300, 13300, 14400 ])
shampoo = np.array([ 1200, 2100, 3550, 1870, 1560, 1890, 1780, 2860, 2100, 2300, 2400, 1800 ])
hidratante = np.array([ 1500, 1200, 1340, 1130, 1740, 1555, 1120, 1400, 1780, 1890, 2100, 1760 ])

# Gráfico 1 - Total de produtos Vendidos por mês - Linha
# Gráfico 2 - Gráfico com todos os produtos vendidos por mês - Linha
# Gráfico 3 - Comparativo de Creme Facial com Limpeza Facial por mês - Barras
# Gráfico 4 - Stack plot com o total de produtos vendidos a cada mês com todos os produtos
# Gráfico 5 - Pizza com % da quantidade produtos vendidos no ano em cada produto
```

```python
# Gráfico 1

total_vendas = creme_facial + limpeza_facial + pasta_dentaria + sabonete + shampoo + hidratante

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(mes, total_vendas, marker='o', color='b', linewidth=2)
ax.set_title('Gráfico 1 - Total de Produtos Vendidos por Mês')
ax.set_xlabel('Mês')
ax.set_ylabel('Total de Vendas')
ax.set_xticks(mes)
ax.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# Gráfico 2

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(mes, creme_facial, label='Creme Facial', marker='o')
ax.plot(mes, limpeza_facial, label='Limpeza Facial', marker='o')
ax.plot(mes, pasta_dentaria, label='Pasta Dentária', marker='o')
ax.plot(mes, sabonete, label='Sabonete', marker='o')
ax.plot(mes, shampoo, label='Shampoo', marker='o')
ax.plot(mes, hidratante, label='Hidratante', marker='o')

ax.set_title('Gráfico 2 - Vendas de Todos os Produtos por Mês')
ax.set_xlabel('Mês')
ax.set_ylabel('Unidades Vendidas')
ax.set_xticks(mes)
ax.legend(loc='upper left')
ax.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#Grafico 3

fig, ax = plt.subplots(figsize=(10, 5))
largura = 0.35  # deslocamento para desenhar as barras lado a lado

ax.bar(mes - largura/2, creme_facial, width=largura, label='Creme Facial')
ax.bar(mes + largura/2, limpeza_facial, width=largura, label='Limpeza Facial')

ax.set_title('Gráfico 3 - Comparativo: Creme Facial vs Limpeza Facial')
ax.set_xlabel('Mês')
ax.set_ylabel('Unidades Vendidas')
ax.set_xticks(mes)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
#Grafico 4

fig, ax = plt.subplots(figsize=(10, 5))
ax.stackplot(mes, creme_facial, limpeza_facial, pasta_dentaria, sabonete, shampoo, hidratante,
             labels=['Creme Facial', 'Limpeza Facial', 'Pasta Dentária', 'Sabonete', 'Shampoo', 'Hidratante'])

ax.set_title('Gráfico 4 - Stack Plot do Total de Vendas por Mês')
ax.set_xlabel('Mês')
ax.set_ylabel('Unidades Totais')
ax.set_xticks(mes)
ax.legend(loc='upper left')
plt.tight_layout()
plt.show()
```

```python
#Grafico 5

fig, ax = plt.subplots(figsize=(8, 8))

totais = [
    creme_facial.sum(),
    limpeza_facial.sum(),
    pasta_dentaria.sum(),
    sabonete.sum(),
    shampoo.sum(),
    hidratante.sum()
]
labels = ['Creme Facial', 'Limpeza Facial', 'Pasta Dentária', 'Sabonete', 'Shampoo', 'Hidratante']

ax.pie(totais, labels=labels, autopct='%1.1f%%', startangle=140)
ax.set_title('Gráfico 5 - Participação % Anual por Produto')
ax.axis('equal')  # Garante formato circular
plt.tight_layout()
plt.show()
```
